```python
import numpy as np



# Normal connections: [packets/sec, distinct ports]

normal = np.array([

    [2, 1], [3, 2], [2, 2], [3, 1], [2, 1], [3, 2]

])



centre = normal.mean(axis=0)          # the "average normal" vector

dists = np.linalg.norm(normal - centre, axis=1)

threshold = dists.max() * 1.5         # simple cutoff



def is_anomaly(x):

    d = np.linalg.norm(np.array(x) - centre)

    return d > threshold



print(centre, threshold)

print(is_anomaly([2, 1]))

print(is_anomaly([5, 5]))

print(is_anomaly([20, 10]))
```
