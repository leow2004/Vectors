```python
import numpy as np

A = np.array([2,1])
B = np.array([5,5])
C = np.array([3,2])
D = np.array([20,10])

def distance(x, y):
  return np.linalg.norm(x-y)

def cosine(x, y):
  return x @ y / (np.linalg.norm(x) * np.linalg.norm(y))

print(distance(A, B), cosine(A, B))
print(distance(B, C), cosine(B, C))
print(distance(C, A), cosine(C, A))
print(distance(D, A), cosine(D, A))

def find_most_similar(vector_pairs):
    """
    Finds the pair of vectors with the cosine similarity closest to 1
    and correlates them to threat intelligence attack vectors.
    """
    best_pair = None
    best_similarity = -1.0

    for label, (v1, v2) in vector_pairs.items():
        sim = cosine(v1, v2)
        print(f"Similarity of {label}: {sim:.5f}")
        if sim > best_similarity:
            best_similarity = sim
            best_pair = label

    print(f"\nMost similar pair is '{best_pair}' with a cosine similarity of {best_similarity:.5f}.")
    print(f"[ASSUMPTION] Highly correlated indicators detected. Because these threat vectors are very similar, ")
    print(f"             it is highly probable that the underlying attack vector was similar or coordinated.")
    return best_pair, best_similarity

# Define our pairs
pairs = {
    'A and B': (A, B),
    'B and C': (B, C),
    'C and A': (C, A),
    'A and D': (A, D),
    'D and A': (D, A),
    'B and D': (B, D),
    'D and B': (D, B),
    'C and D': (C, D),
    'D and C': (D, C)
}

# Run the function
_ = find_most_similar(pairs)
```
